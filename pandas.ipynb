```python
import pandas as pd
```

```python
df=pd.read_csv('Housing.csv')
print(df)
```

```python
area=df['area']
price=df['price']
print(area.head())
print(price.head())
```

```python
print(price.count())
print(area.count())
```

```python
def remove_null(data):
    if data.isnull().any():
        data = data.dropna()  
    return data

price = remove_null(price)
area = remove_null(area)
```

```python
new_df=pd.DataFrame({
    "area": area,
    'price': price
})
new_df.head()
```

```python
new_df.to_csv('clean_housing_area_price.csv', index=False)
```
