```python

# creating a sample array (dummy data)
arr = np.array([10, 20, 30, 40, 50])

# print original data
print("Array values:", arr)

# check total elements
print("Array size:", arr.size)

# check array shape
print("Array shape:", arr.shape)

# sum of all values
total = np.sum(arr)
print("Sum:", total)

# average using average()
avg = np.average(arr)
print("Average using np.average:", avg)

# mean using mean()
m = np.mean(arr)
print("Mean using np.mean:", m)

# max and min values
print("Maximum value:", np.max(arr))
print("Minimum value:", np.min(arr))

# standard deviation
print("Standard deviation:", np.std(arr))
```
